```python
import pandas as pd
import numpy as np
```

```python
tEmployees = pd.read_csv('tEmployees.csv')
tStores = pd.read_csv('tStores.csv')
tSales = pd.read_csv('tSales.csv')
```

```python
#aggregate function
# first we pick our grouping then we pick what to aggregate
# dataframename.groupby(specify the name of the grouping column).agg(
# name of the new column = (column upon which to aggregate, aggregation function)
# )
# aggregation functions include sum, count, min, max, mean, median

cityStores = tStores.groupby('City').agg(
    sumSqFt = ('SqFt', 'sum'),
    numStores = ('Store', 'count')
)

#1) tStores – count number of stores in each city, sum square footage. Filter for more than one store
cityStores.loc[cityStores['numStores']>1]
```

```python
prices_table = pd.read_csv('merge_prices.csv')
qty_table = pd.read_csv('merge_qty.csv')
```

```python
#2) Temployees – total number of employees by city > I will need to merge tStores to
# tEmployees to find the city each employee works in

#assign a variable so we can continue using it

cityEmployees = pd.merge(tEmployees, tStores, on='Store', how = 'left')
```

```python
# then aggregate total employees based on city
cityEmployees.groupby('City').agg(
    totalEmployees = ('emplID', 'count')
)
```

```python
#3) tSales – merge with tStores to figure out which city each transaction took place in
Store_CitySales = pd.merge(tSales, tStores, on='Store', how = 'left')
```

```python
#4) Sales*quantity = revenue per transaction
Store_CitySales['Revenue'] = Store_CitySales['SalesPrice'] * Store_CitySales['SalesQuantity']

tSales_Summary = (Store_CitySales.groupby('City').agg(totalRevenue = ('Revenue', 'sum')))
```
